```python
import pandas as pd
data = pd.read_csv("robot_mission.csv")
numbers = data["number"].tolist()
rounds = data["round"].tolist()
print(numbers)
print(rounds)
```

```python
total= 0
positions = []
for n in range(len(numbers)):
  i = numbers[n]
  r = i%2
  if r==0:
    total += i
    positions.append(total)
  else:
    total -= i
    positions.append(total)
  print("Current Number: ", i)
  print("Current Position:", positions[n])
```
